```python
import math
def sigmoid(x):
    return 1 / (1 + math.exp(-x))

```

```python
sigmoid(250)
```

```python
sigmoid(-556)
```

```python
def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))
```

```python
tanh(250)
```

```python
tanh(-556)
```

```python
def relu(x):
    return max(0, x)
```

```python
relu(250)
```

```python
relu(-250)
```

```python
def leaky_relu(x, alpha=0.01):
    return x if x > 0 else alpha * x
```

```python
leaky_relu(250)
```

```python
leaky_relu(-250)
```
